## SPLITTING DATA

```python
x = df.drop(columns=['ID_code', 'target'])
y = df['target']
```

## DATA PREPROCESSING

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

## FEATURE SCALING

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train
```

```python
y_train
```

## MODEL BUILDING

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

## LOGISTIC REGRESSION

```python
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
y_pre=model.predict(x_test)
```

```python
y_pre
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
print("Accuracy :", accuracy_score(y_test, y_pre))
print("Precision:", precision_score(y_test, y_pre))
print("Recall   :", recall_score(y_test, y_pre))
print("F1 Score :", f1_score(y_test, y_pre))
```

## DECISION TREE

```python
dt_model = DecisionTreeClassifier()
dt_model.fit(x_train, y_train)
```

## SVM

```python
svm_model = LinearSVC()
svm_model.fit(x_train, y_train)
```

```python
y_pre_svm = svm_model.predict(x_test)
```

```python
print("SVM Results") 
print("Accuracy :", accuracy_score(y_test, y_pre_svm))
print("Precision:", precision_score(y_test, y_pre_svm))
print("Recall   :", recall_score(y_test, y_pre_svm))
print("F1 Score :", f1_score(y_test, y_pre_svm))
```

## RANDOM FOREST

```python
rf_model = RandomForestClassifier(n_jobs=-1,random_state=42)
rf_model.fit(x_train, y_train)
```

```python
y_pre_rf = rf_model.predict(x_test)
```

```python
print("Random Forest Results")
print("Accuracy :", accuracy_score(y_test, y_pre_rf))
print("Precision:", precision_score(y_test, y_pre_rf))
print("Recall   :", recall_score(y_test, y_pre_rf))
print("F1 Score :", f1_score(y_test, y_pre_rf))
```

## WITH TUNING

## Importing

```python
from sklearn.metrics import classification_report
from sklearn.model_selection import GridSearchCV
```

## LOGISTIC REGRESSION

```python
param_grid={
    "penalty":["l2"],
    "C":[0.1],
    "solver":["liblinear"]
}
```

```python
lg_model=LogisticRegression(max_iter=400)

lg_grid=GridSearchCV(lg_model,param_grid,cv=5)
lg_grid.fit(x_train,y_train)
```

```python
log_best = lg_grid.best_estimator_
log_tuned_pre = log_best.predict(x_test)
```

```python
log_tuned_pre
```

```python
lg_tuned_accuracy = accuracy_score(y_test, log_tuned_pre)
lg_tuned_accuracy
```

## DECISION TREE

```python
dt_params = {
    "criterion": ["gini"],
    "max_depth": [10,15],
    "min_samples_split": [2,5,10]
}
```

```python
dt_model = DecisionTreeClassifier()
dt_grid = GridSearchCV(dt_model, dt_params, cv=5)
dt_grid.fit(x_train, y_train)
```

```python
dt_best = dt_grid.best_estimator_
dt_tuned_pred = dt_best.predict(x_test)
dt_tuned_accuracy = accuracy_score(y_test, dt_tuned_pred)
dt_tuned_accuracy 
```

## SVM

```python
svm_model = LinearSVC(max_iter=5000)
svm_params = {
    "C": [0.01, 0.1, 1, 10]
}
```

```python
svm_grid = GridSearchCV(svm_model, svm_params, cv=3)
svm_grid.fit(x_train, y_train)
```

```python
svm_best = svm_grid.best_estimator_
svm_tuned_pred = svm_best.predict(x_test)
```

```python
svm_tuned_accuracy = accuracy_score(y_test, svm_tuned_pred)
svm_tuned_accuracy
```

## RANDOM FOREST

```python
rf_params = {
    "n_estimators": [100],
    "max_depth": [10,20],
    "min_samples_split": [5]
}
```

```python
rf_model = RandomForestClassifier()

rf_grid = GridSearchCV(rf_model, rf_params, cv=5)
rf_grid.fit(x_train, y_train)
```

```python
rf_best = rf_grid.best_estimator_
rf_tuned_pred = rf_best.predict(x_test)
```

```python
rf_tuned_accuracy = accuracy_score(y_test, rf_tuned_pred)
rf_tuned_accuracy
```

## XGBOOST

```python
xgb_params = {
    "n_estimators": [100],
    "max_depth": [5],
}
```

```python
xgb = GridSearchCV(
    XGBClassifier(
        objective="binary:logistic",
        eval_metric="logloss",
        use_label_encoder=False,
        random_state=42
    ),
    xgb_params,
    cv=3
)

xgb.fit(x_train, y_train)
```

```python
xgb_best = xgb.best_estimator_
xgb_tuned_pred = xgb_best.predict(x_test)
xgb_tuned_accuracy = accuracy_score(y_test, xgb_tuned_pred)
xgb_tuned_accuracy
```

```python
print("ACCURACY FOR ALL MODELS WITH TUNNING")
print("LOGISTIC REGRESSION :", lg_tuned_accuracy)
print("DECISION TREE :",dt_tuned_accuracy)
print("SVM:", svm_tuned_accuracy)
print("RANDOM FOREST:",rf_tuned_accuracy)
print("XGB BOOST:", xgb_tuned_accuracy)
```
